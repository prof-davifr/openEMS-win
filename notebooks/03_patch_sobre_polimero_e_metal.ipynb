# 3. O sensor sobre a amostra: polímero sobre metal

Agora a antena fica voltada para baixo, sobre um revestimento de polímero aplicado num tubo de aço:

```
        plano de terra da antena
        substrato FR-4 (1,6 mm)
        patch  (voltado para baixo)
        ---- gap de ar ----
        polímero (espessura t, er ~ 3-5)
z = 0   metal (aço = condutor perfeito)
```

A classe `emslab.Simulacao` monta esta geometria com as mesmas etapas do notebook 02.
O código dela está em `emslab/src/emslab/simulacao.py`: leia quando quiser ver os detalhes.

````python
import matplotlib.pyplot as plt
import emslab

antena = emslab.patch_retangular(f_ghz=3.5, er=4.4, h=1.6, tand=0.02)
print(antena)
````

## 3.1 Antena sozinha, no ar

````python
sim = emslab.Simulacao(f_min=2, f_max=5)
sim.antena(antena)
no_ar = sim.rodar("sensor_no_ar")
````

## 3.2 Antena sobre a amostra

Parâmetros da amostra: espessura do polímero (mm), permissividade, tangente de perdas e distância de ar até o patch (mm).

````python
sim = emslab.Simulacao(f_min=2, f_max=5)
sim.antena(antena)
sim.amostra(espessura=10, er=4.0, tand=0.01, gap=3)
sim.ver("sensor_sobre_amostra")     # abre a geometria em 3D; feche a janela para continuar
````

````python
sobre_amostra = sim.rodar("sensor_sobre_amostra")
````

## 3.3 Comparação

````python
emslab.grafico_s11(no_ar, sobre_amostra)
plt.savefig("03_s11_ar_e_amostra.png", dpi=150)

delta = (sobre_amostra.f_ressonancia - no_ar.f_ressonancia) * 1000
print(f"A amostra deslocou a ressonância em {delta:+.0f} MHz")
````
