## Variáveis e contas

As unidades não existem para o Python. Neste projeto, usamos **mm** para tamanhos e **GHz** para frequência.

```python
f = 3.5                 # frequência em GHz
c = 299.792458          # velocidade da luz em mm·GHz
comprimento_de_onda = c / f
print(f"Em {f} GHz, o comprimento de onda no ar é {comprimento_de_onda:.1f} mm")
```

## Listas, laços e o numpy

O `numpy` faz contas com muitos números de uma vez.

```python
import numpy as np

er = np.array([1, 2.2, 3, 4, 4.4, 5])        # permissividades relativas
lambda_no_material = c / f / np.sqrt(er)
for e, l in zip(er, lambda_no_material):
    print(f"er = {e:4}: comprimento de onda = {l:5.1f} mm")
```

## Gráficos com o matplotlib

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(er, lambda_no_material, "o-")
ax.set_xlabel("Permissividade relativa")
ax.set_ylabel("Comprimento de onda (mm)")
ax.set_title(f"Comprimento de onda em {f} GHz")
ax.grid(True)
fig.savefig("comprimento_de_onda.png", dpi=150)   # grava a figura na pasta deste notebook
```
